```python
import cv2 as cv
import matplotlib.pyplot as plt

image = cv.imread('data/road/misty_road.jpeg')

gray = cv.cvtColor(image, cv.COLOR_BGR2GRAY)
plt.imshow(gray, cmap='gray'), plt.xticks([]), plt.yticks([]), plt.show()

histogram = cv.calcHist([gray], [0], None, [256], [0,256])
plt.plot(histogram, color='r', linewidth=1), plt.show()

after_equalize = cv.equalizeHist(gray)
plt.imshow(after_equalize, cmap='gray'), plt.xticks([]), plt.yticks([]), plt.show()

histogram = cv.calcHist([after_equalize], [0], None, [256], [0,256])
plt.plot(histogram, color='r', linewidth=1), plt.show()
```
