```python
# 이미지에서 이진화를 위한 임계값 설정을 위해 히스토그램 구하기
import cv2 as cv
import matplotlib.pyplot as plt

image = cv.imread('data/soccer/soccer.jpg')
# 2번 채널인 R 채널에서 히스토그램 구함
# calcHist : 인수를 리스트로 주어야 함
# calcHist([이미지], [이미지 채널 번호], '히스토그램을 구할 영역을 지정하는 마스크', [히스토그램 칸 수], [세어볼 명암값의 범위])
histogram = cv.calcHist([image], [2], None, [256], [0,256])
plt.plot(histogram, color='r', linewidth=1)
```

```python
# 임계값을 정하는 일을 쉽게 하기 위한 방법론
# Otsu Algorithm
# => 모든 명암값에 대해 목적함수 J를 계산하고 J가 최소인 명암값을 최적값 t로 정함
# => 이렇게 결정한 t를 임계값 T로 사용해 이진화

# Otsu Algorithm으로 이진화하기
import cv2 as cv
import sys

image = cv.imread('data/soccer/soccer.jpg')

# cv.THRESH_OTSU => Otsu Algorithm을 활용하여 이진화 수행
t, bin_image = cv.threshold(image[:,:,2], 0, 255, cv.THRESH_BINARY + cv.THRESH_OTSU)
print('optimum threshold by ostu algorithm = ', t)

# R 채널 이미지
cv.imshow('R channel', image[:,:,2])

# R 채널 이진화 이미지
cv.imshow('R channel binarization', bin_image)

cv.destroyAllWindows()
```
